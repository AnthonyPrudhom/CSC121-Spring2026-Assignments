```python
# IMPORT STATEMENTS
# noen

# FUNCTION DEFINITIONS
# 1. define funciton is_palindrome(word)
def is_palindrome(word):
# 2. convert all characrers in input to lower case
  word = word.lower()
# 3. reverse input string and store in new variable
  reversed_word = word[::-1]
# 4. compare to original input abd return true or false if palendrom
  if word == reversed_word:
    return True
  else:
    return False


# PROGRAM LOGIC
# 1. input() for word
# 2. call is_palindrome(word)
is_palindrome("Skeleton")
is_palindrome("Racecar")

```
