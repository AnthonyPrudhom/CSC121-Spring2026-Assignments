```python
# NAME: gradebook.py
# IMPORT STATEMENTS
#none

# FUNCTION DEFINITIONS
# 1. define add_students(dict)
def add_students(dict):
# name = input("Enter student name (or 'done'): ")
  name = input("Enter student name (or 'done'): ")
# 2. while name != "done", grade = float(input())
  while name != "done":
    grade = float(input("Enter " + name + "'s grade: "))
# 3. dict[name] = grade
    dict[name] = grade
# 4. name = input("Enter student name (or 'done'): ")
    name = input("Enter student name (or 'done'): ")
# 5. define calculate_average(dict)
def calculate_average(dict):
# 6. grades = dict.values()
  grades = dict.values()
# 7. return sum(grades)/len(grades)
  if len(grades) == 0:
    return 0
  else :
    return sum(grades)/len(grades)


# PROGRAM LOGIC
# define gradebook
gradebook = {}
# call add_students(gradebook)
add_students(gradebook)
# define avg = alculate_average(gradebook)
avg = calculate_average(gradebook)
# print("--- Class Report ---")
print("--- Class Report ---")
# for name in gradebook, print(name + ": " + gradebook(name)
for name in gradebook:
  print(name + ": " + str(gradebook[name]))
# print("Class Average: " + calculate_average(gradebook) )
print("Class Average:" + f"{avg: .2f}")


```
