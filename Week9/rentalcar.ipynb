```python
# NAME: rentalcar.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS.
# 1. define calc_total(Car, startODE, endODE, days)
def calc_total(Car, startODE, endODE, days):
# 2.check if entered car type is one of three to get daily rate (B = $40/day, S = $60/day,  L = $80/day)
  if Car == "B":
    rate = 40
  elif Car == "S":
    rate = 60
  else:
    rate = 80
# 4.if number of miles is less than 100, do not include miles driven * rate.
  if (endODE - startODE) <= 100:
    total = days*rate
# 5.if over 100 miles driven Comebine product of days rented* rate plus amount of miles driver -100 times $0.35/mile
  else:
    total = (endODE - startODE -100)*0.35 + days*rate
# 6. Return total bill
  return total


# PROGRAM LOGIC
# 1.get user inputs for the Car, StartODE, endODE, days
Car = input("Enter classification code: ")
while Car not in ["B", "S", "L"]:
    print("Invalid code. Please enter B, S, or L.")
    Car = input("Enter classification code (B, S, L): ")

startODE = int(input("Start Odometer: "))
endODE = int(input("End Odometer: "))
days = int(input("Days Rented:"))
# 2. call calc_total(calc_total(Car, startODE, endODE, days)
total = calc_total(Car, startODE, endODE, days)
# 3. store difference of endODE and startODE in variable: total_Miles
driven_miles = endODE-startODE
# 4. if driven miles is over 100 in variable: miles_Charged = 0, else miles charged = driven_miles - 100
if (driven_miles) <= 100:
  charged_miles = 0
else:
  charged_miles = driven_miles - 100
# 5. if car input: B, car = budget, if S, car = Standard, if L, car = luxury.
if Car == "B":
  classification = "Budget"
elif Car == "S":
  classification = "Standard"
else:
  classification = "Luxury"
# 5. print classification, total miles driven, miles charged, total bill,
print("Classification: " + classification)
print("Total Miles: " + str(driven_miles))
print("Charged Miles: " + str(charged_miles))
print("Total Bill: $" + str(total))
```
