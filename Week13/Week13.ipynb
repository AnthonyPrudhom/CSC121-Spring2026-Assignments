```python
# NAME: journal.py
# IMPORT STATEMENTS
import os

# FUNCTION DEFINITIONS
# 1.define write_entry(file,entry)
def write_entry(filename,entry):
# 2.open the file in append mode
  f = open(filename,"a")
# 3.prompt user for input and append input to file
  f.write(entry + "\n")
  print("Entry saved")
# 4.close the file
  f.close()
# 5.define read_journal(file)
def read_journal(filename):
# 7.check if empty by seeing file path exists
  if os.path.exists(filename):
  # 8.if it exists open file and print contents
    f = open(filename,"r")
    print("---Your Jounral---")
    print(f.read())
    # 9. close file
    f.close()
  # 10. if not then print empty journal message
  else:
    print("Error file not found")


# PROGRAM LOGIC
#11. Define jounral.txt
filename = "journal.txt"
#12. prompt user for either W or R their journal
option = input("W)rite, (R)ead, or (Q)uit?")
#13. while input != Q
while option.upper != "Q":
# if W store user input and write user input to filename
  if option.upper() == "W":
    entry = input("Enter your entry:")
    write_entry(filename,entry)
# if R, read the journal
  elif option.upper() == "R":
    read_journal(filename)
# if Q then break
  elif option.upper() == "Q":
    break
# otherwise print invalid option message
  else:
    print("Invalid option")
# update user inupt
  option = input("W)rite, (R)ead, or (Q)uit?")



```
