```python
# NAME: rectangle.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
#1. Define calculate_rectangle(length, width)
def calculate_rectangle(length, width):
#2. define Area =(length * width)
  Area =(length * width)
#3. define Perimeter =(2 * (length + width))
  Perimeter =(2 * (length + width))
#4. define t = Area, Perimeter tuple
  t = (Area, Perimeter)
#5. return (t)
  return t


# PROGRAM LOGIC
#6. Define Width = float user input('Enter width: ')
Width = float(input('Enter width: '))
#7. Define Length = float user input('Enter length: ')
Length = float(input('Enter length: '))
#8. Define returned tuple = calling calculate_rectangle(Length, Width)
returned_tuple = calculate_rectangle(Length, Width)
#9. define tuple my_area, my_perimeter = returned tuple from calling calculate_rectangle
my_area, my_perimeter = returned_tuple
#10. Print results message
print("---Results---")
#11. print area result message
print("The Area is: " + str(my_area))
#12. print Perimeter reuslt message
print("The Perimeter is: " + str(my_perimeter))





```
